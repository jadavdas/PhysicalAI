```python
import pybullet
import pybullet_data
```

```python
physics_client = pybullet.connect(pybullet.GUI) # GUI, DIRECT, SHARED_MEMORY, UDP, TCP
```

```python
# Reset the space
pybullet.resetSimulation()
# Add paths for the necessary data
pybullet.setAdditionalSearchPath(pybullet_data.getDataPath())
# Set gravity
pybullet.setGravity(0, 0, -9.81)
time_step = 1.0/240
pybullet.setTimeStep(time_step)
```

```python
# Load the floor
plane_id = pybullet.loadURDF('plane.urdf')
```

```python
mass=10
box_size=[0.5,0.5,0.5]
position=[2,0,0.3]
orientation=[0,0,0,1] #quaternion
collision_shape_id=pybullet.createCollisionShape(pybullet.GEOM_BOX,halfExtents=box_size)
visual_shape_id=pybullet.createVisualShape(pybullet.GEOM_BOX,halfExtents=box_size,rgbaColor=[1,0,0,1])
box_id=pybullet.createMultiBody(mass,collision_shape_id,visual_shape_id,position,orientation)
```
